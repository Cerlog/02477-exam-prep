# Section 8 — Mixture Models

Mixture models represent data as coming from a combination of several component distributions.

---

## 8.1 Gaussian Mixture Models (GMM)

A **Gaussian Mixture Model** assumes:
$$
p(\mathbf{x}) = \sum_{k=1}^K \pi_k \, \mathcal{N}(\mathbf{x} \mid \boldsymbol{\mu}_k, \Sigma_k)
$$
where:
- $\pi_k \ge 0$, $\sum_{k=1}^K \pi_k = 1$ (mixing proportions)
- $\mathcal{N}(\mathbf{x} \mid \boldsymbol{\mu}_k, \Sigma_k)$ is the $k$-th Gaussian component.

---

### 8.1.1 Latent Variable Formulation

Introduce latent variable $z \in \{1,\dots,K\}$:
- $p(z=k) = \pi_k$
- $p(\mathbf{x} \mid z=k) = \mathcal{N}(\mathbf{x} \mid \boldsymbol{\mu}_k, \Sigma_k)$

Then:
$$
p(\mathbf{x}) = \sum_{k=1}^K p(z=k) \, p(\mathbf{x} \mid z=k)
$$

---

### 8.1.2 Parameter Estimation via EM Algorithm

We maximize $\log p(\mathbf{X} \mid \pi, \mu, \Sigma)$ iteratively:

**E-step:**
$$
r_{nk} = p(z_n = k \mid \mathbf{x}_n) = \frac{\pi_k \, \mathcal{N}(\mathbf{x}_n \mid \mu_k, \Sigma_k)}{\sum_{j=1}^K \pi_j \, \mathcal{N}(\mathbf{x}_n \mid \mu_j, \Sigma_j)}
$$

**M-step:**
$$
N_k = \sum_{n=1}^N r_{nk}
$$
$$
\pi_k = \frac{N_k}{N}
$$
$$
\mu_k = \frac{1}{N_k} \sum_{n=1}^N r_{nk} \, \mathbf{x}_n
$$
$$
\Sigma_k = \frac{1}{N_k} \sum_{n=1}^N r_{nk} \, (\mathbf{x}_n - \mu_k)(\mathbf{x}_n - \mu_k)^\top
$$

Repeat until convergence.

---

### 8.1.3 Bayesian GMM

- Place priors on $\pi_k$ (Dirichlet) and Gaussian parameters.
- Inference: Gibbs sampling, variational inference.
- Nonparametric version: **Dirichlet Process Gaussian Mixture Model** (DP-GMM) — allows $K$ to be infinite.

---

## 8.2 Exam Tips

- Always write GMM as **sum of weighted Gaussians**.
- If EM is required: state E-step (responsibilities) and M-step formulas clearly.
- For Bayesian GMM: mention priors and inference methods (e.g., Gibbs).
- Interpret $\pi_k$: proportion of data from component $k$.
