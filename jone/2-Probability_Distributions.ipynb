# Section 2 — Probability Distributions

This section is your **distribution toolbox**. These appear repeatedly in priors, likelihoods, posteriors, and predictive formulas.  
We’ll focus on **definitions**, **key properties**, and **Bayesian conjugacy**.

---

## 2.1 Beta Distribution

**Support:** $\theta \in (0,1)$

**PDF:**
$$
\mathrm{Beta}(\theta \mid \alpha,\beta) \;=\; \frac{\theta^{\alpha-1}\,(1-\theta)^{\beta-1}}{B(\alpha,\beta)}, \quad
B(\alpha,\beta) = \frac{\Gamma(\alpha)\,\Gamma(\beta)}{\Gamma(\alpha+\beta)}.
$$

**Properties:**
- Mean: $\; \mathbb{E}[\theta] = \frac{\alpha}{\alpha+\beta}$
- Variance: $\; \mathrm{Var}[\theta] = \frac{\alpha\beta}{(\alpha+\beta)^2\,(\alpha+\beta+1)}$

**Conjugacy:**
- Prior: $\mathrm{Beta}(\alpha,\beta)$  
- Likelihood: Binomial$(N,\theta)$  
- Posterior: $\mathrm{Beta}(\alpha+y,\,\beta+N-y)$

---

## 2.2 Binomial Distribution

**Support:** $y \in \{0,1,\dots,N\}$

**PMF:**
$$
\mathrm{Bin}(y \mid N,\theta) = \binom{N}{y} \,\theta^y\,(1-\theta)^{N-y}.
$$

**Properties:**
- Mean: $\; \mathbb{E}[y] = N\theta$
- Variance: $\; \mathrm{Var}[y] = N\theta(1-\theta)$

**Use:** Likelihood for count of successes in $N$ trials.

---

## 2.3 Beta–Binomial Distribution

Arises from **marginalizing $\theta$** in the Binomial likelihood with a Beta prior.

**Model:**  
$\theta \sim \mathrm{Beta}(\alpha,\beta)$,  
$y \mid \theta \sim \mathrm{Bin}(N,\theta)$.

**PMF:**
$$
p(y) = \binom{N}{y} \frac{B(\alpha+y,\;\beta+N-y)}{B(\alpha,\beta)}.
$$

**Properties:**
- Mean: $\; \mathbb{E}[y] = N \frac{\alpha}{\alpha+\beta}$
- Variance: $\; \mathrm{Var}[y] = N\frac{\alpha\beta}{(\alpha+\beta)^2} \cdot \frac{\alpha+\beta+N}{\alpha+\beta+1}$

**Interpretation:** Overdispersed relative to plain Binomial.

---

## 2.4 Poisson Distribution

**Support:** $y \in \{0,1,2,\dots\}$

**PMF:**
$$
\mathrm{Pois}(y \mid \lambda) = \frac{\lambda^y\,e^{-\lambda}}{y!}.
$$

**Properties:**
- Mean: $\; \lambda$
- Variance: $\; \lambda$

**Conjugacy:**
- Prior: $\mathrm{Gamma}(\alpha,\beta)$ (shape, rate)
- Posterior: $\mathrm{Gamma}(\alpha+\sum y_i,\,\beta+n)$

---

## 2.5 Gamma Distribution

**Support:** $x > 0$

**PDF (shape–rate):**
$$
\mathrm{Gamma}(x \mid \alpha,\beta) = \frac{\beta^\alpha}{\Gamma(\alpha)}\,x^{\alpha-1}\,e^{-\beta x}.
$$

**Properties:**
- Mean: $\; \frac{\alpha}{\beta}$
- Variance: $\; \frac{\alpha}{\beta^2}$

**Common roles:**
- Prior for positive rates (Poisson $\lambda$, exponential rate)
- Prior for inverse variance (precision) in Gaussian models

---

## 2.6 Gaussian (Normal) Distribution

**Support:** $x \in \mathbb{R}$

**PDF:**
$$
\mathcal{N}(x \mid \mu,\sigma^2) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\!\left( -\frac{(x-\mu)^2}{2\sigma^2} \right).
$$

**Properties:**
- Mean: $\mu$
- Variance: $\sigma^2$

**Conjugacy:**
- Known $\sigma^2$, prior $\mu \sim \mathcal{N}(\mu_0,\tau_0^2)$ $\;\Rightarrow\;$ Gaussian posterior for $\mu$
- Unknown variance: Inverse-Gamma / scaled-Inv-$\chi^2$ priors

---

## 2.7 Multivariate Gaussian

**Support:** $\mathbf{x} \in \mathbb{R}^d$

**PDF:**
$$
\mathcal{N}(\mathbf{x} \mid \boldsymbol\mu, \Sigma) = \frac{1}{\sqrt{(2\pi)^d \,|\Sigma|}} \exp\!\left( -\frac12 (\mathbf{x}-\boldsymbol\mu)^\top \Sigma^{-1} (\mathbf{x}-\boldsymbol\mu) \right).
$$

**Key facts:**
- Marginals and conditionals are Gaussian
- Sum of independent Gaussians is Gaussian
- **Conditioning formula** (partitioned covariance):
  $$
  \mathbf{x}_a \mid \mathbf{x}_b \;\sim\; \mathcal{N}\!\left( \mu_a + \Sigma_{ab} \Sigma_{bb}^{-1} (\mathbf{x}_b - \mu_b), \;\Sigma_{aa} - \Sigma_{ab}\Sigma_{bb}^{-1}\Sigma_{ba} \right).
  $$

---

## 2.8 Dirichlet Distribution

**Support:** $\boldsymbol{\theta} \in \{\theta_k \ge 0, \sum_{k=1}^K \theta_k = 1\}$

**PDF:**
$$
\mathrm{Dir}(\boldsymbol{\theta} \mid \boldsymbol{\alpha}) = \frac{1}{B(\boldsymbol{\alpha})} \prod_{k=1}^K \theta_k^{\alpha_k - 1},
\quad
B(\boldsymbol{\alpha}) = \frac{\prod_{k=1}^K \Gamma(\alpha_k)}{\Gamma(\sum_{k=1}^K \alpha_k)}.
$$

**Properties:**
- Mean: $\; \mathbb{E}[\theta_k] = \frac{\alpha_k}{\alpha_0}, \quad \alpha_0 = \sum_k \alpha_k$
- Variance: $\; \mathrm{Var}[\theta_k] = \frac{\alpha_k(\alpha_0-\alpha_k)}{\alpha_0^2(\alpha_0+1)}$

**Conjugacy:**
- Prior: $\mathrm{Dir}(\boldsymbol{\alpha})$
- Likelihood: Multinomial$(n,\boldsymbol{\theta})$
- Posterior: $\mathrm{Dir}(\alpha_k + y_k)$

---

## 2.9 Wishart Distribution

Distribution over **positive-definite matrices**.

**Support:** $\Sigma \in \mathbb{R}^{d\times d}$, $\Sigma \succ 0$

**PDF:**
$$
\mathcal{W}(\Sigma \mid V, \nu) \propto |\Sigma|^{(\nu-d-1)/2} \exp\!\left( -\frac12 \mathrm{tr}(V^{-1} \Sigma) \right),
$$
where $\nu$ is degrees of freedom, $V$ is scale matrix.

**Properties:**
- Mean: $\;\mathbb{E}[\Sigma] = \nu V$ (for $\nu > d-1$)

**Conjugacy:**
- Prior for precision matrix $\Lambda = \Sigma^{-1}$ in multivariate Gaussian models.

---

## 2.10 Conjugate Model Summary

| Likelihood | Prior | Posterior |
|------------|-------|-----------|
| Binomial$(N,\theta)$ | Beta$(\alpha,\beta)$ | Beta$(\alpha+y,\,\beta+N-y)$ |
| Poisson$(\lambda)$ | Gamma$(\alpha,\beta)$ | Gamma$(\alpha+\sum y_i,\,\beta+n)$ |
| Gaussian mean (known var) | Gaussian | Gaussian |
| Multinomial$(n,\boldsymbol{\theta})$ | Dirichlet | Dirichlet$(\alpha_k+y_k)$ |
| Gaussian (unknown var) | Normal–Inverse-Gamma | Normal–Inverse-Gamma |

---

## 2.11 Exam Tip
- Always **name the distribution** and write its **parameters clearly**.  
- For conjugate cases, write **posterior parameters directly** — don’t rederive unless asked.  
- Remember: **marginal likelihood** in conjugate pairs often has a closed form (useful for model comparison questions).
