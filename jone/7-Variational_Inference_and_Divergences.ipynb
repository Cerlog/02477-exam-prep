# Section 7 — Variational Inference & Divergences

Variational Inference (VI) approximates an intractable posterior $p(\theta \mid y)$ with a tractable distribution $q(\theta)$ by **optimization**.

---

## 7.1 KL Divergence

The **Kullback–Leibler divergence** between distributions $q(\theta)$ and $p(\theta)$ is:
$$
\mathrm{KL}(q \,\|\, p) = \int q(\theta) \log \frac{q(\theta)}{p(\theta)} \, d\theta
$$

**Properties:**
- Non-negative: $\mathrm{KL}(q \,\|\, p) \ge 0$
- Asymmetric: $\mathrm{KL}(q \,\|\, p) \neq \mathrm{KL}(p \,\|\, q)$
- In VI, we often minimize $\mathrm{KL}(q(\theta) \,\|\, p(\theta \mid y))$

---

## 7.2 ELBO (Evidence Lower Bound)

From Bayes' rule:
$$
\log p(y) = \mathcal{L}(q) + \mathrm{KL}(q(\theta) \,\|\, p(\theta \mid y))
$$
where:
$$
\mathcal{L}(q) = \mathbb{E}_q[\log p(y,\theta)] - \mathbb{E}_q[\log q(\theta)]
$$

**Maximizing ELBO** $\mathcal{L}(q)$ $\Leftrightarrow$ minimizing KL divergence to the posterior.

---

## 7.3 Coordinate Ascent VI (CAVI)

**Setting:** $q(\theta) = \prod_{j=1}^m q_j(\theta_j)$ (mean-field assumption)

**Update rule:**
$$
\log q_j^\star(\theta_j) = \mathbb{E}_{q_{-j}}[\log p(y,\theta)] + \text{const}
$$
- Update each $q_j$ in turn, keeping others fixed.
- Often yields closed-form updates for conjugate models.

---

## 7.4 Black Box VI (BBVI)

**Goal:** Apply VI to models without analytic updates.

- Represent $q_\lambda(\theta)$ with parameters $\lambda$
- Optimize ELBO via stochastic gradients:
$$
\nabla_\lambda \mathcal{L}(\lambda) = \mathbb{E}_{q_\lambda} \big[ \nabla_\lambda \log q_\lambda(\theta) \, (\log p(y,\theta) - \log q_\lambda(\theta)) \big]
$$
- Estimate expectation with Monte Carlo samples.

---

## 7.5 α-divergence

Generalization of KL divergence:
$$
D_\alpha(p \,\|\, q) = \frac{1}{\alpha(1-\alpha)} \left( 1 - \int p(\theta)^\alpha q(\theta)^{1-\alpha} \, d\theta \right)
$$
- $\alpha \to 0$: recovers $\mathrm{KL}(q \,\|\, p)$
- $\alpha \to 1$: recovers $\mathrm{KL}(p \,\|\, q)$
- Choice of $\alpha$ controls **mode-seeking** vs **mass-covering** behavior.

---

## 7.6 Score Function Gradient Estimator

Also called **REINFORCE** or **likelihood-ratio trick**.

For $\mathbb{E}_{q_\lambda}[f(\theta)]$:
$$
\nabla_\lambda \mathbb{E}_{q_\lambda}[f(\theta)] = \mathbb{E}_{q_\lambda}[f(\theta) \nabla_\lambda \log q_\lambda(\theta)]
$$
- Unbiased, but high variance.
- Variance reduction via control variates.

---

## 7.7 Log-Derivative Trick

Key identity used above:
$$
\nabla_\lambda q_\lambda(\theta) = q_\lambda(\theta) \nabla_\lambda \log q_\lambda(\theta)
$$
- Allows gradients to pass inside expectation.

---

## 7.8 Reparameterisation Trick

If $\theta = g_\lambda(\epsilon)$ with $\epsilon \sim p(\epsilon)$ independent of $\lambda$:
$$
\nabla_\lambda \mathbb{E}_{q_\lambda}[f(\theta)] = \mathbb{E}_{p(\epsilon)}\left[ \nabla_\lambda f(g_\lambda(\epsilon)) \right]
$$
- Lower variance than score function estimator.
- Example: Gaussian $q_\lambda$: $\theta = \mu + \sigma \epsilon$, $\epsilon \sim \mathcal{N}(0,1)$.

---

## 7.9 Expectation Propagation (EP)

**Goal:** Approximate posterior $p(\theta \mid y) \propto \prod_{i} f_i(\theta)$  
by $q(\theta) \propto \prod_{i} t_i(\theta)$ where each $t_i$ is in an exponential family.

**Procedure:**
1. Remove one factor $t_i$ (forming cavity distribution)
2. Incorporate exact $f_i$
3. Project back onto exponential family (minimizing $\mathrm{KL}(f_i \cdot \text{cavity} \,\|\, t_i \cdot \text{cavity})$)
4. Iterate over all factors.

**Notes:**
- EP minimizes **reverse KL** locally.
- Often better calibrated than VI.

---

## 7.10 Exam Tips

- ELBO derivation: start from $\log p(y)$ and insert $q(\theta)$.
- CAVI: state mean-field assumption and expectation form update.
- BBVI: write score function estimator form.
- Reparameterisation: show $\theta = \mu + \sigma \epsilon$ example.
- EP: emphasize factor replacement & projection step.
