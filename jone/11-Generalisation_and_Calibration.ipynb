# Section 11 — Generalisation & Calibration

This section covers how well models perform on new, unseen data (**generalisation**) and how well predicted probabilities match observed outcomes (**calibration**).

---

## 11.1 Model Generalisation

**Goal:** Ensure the model performs well on unseen data, not just the training set.

### Key Concepts
- **Overfitting**: Model fits noise in training data → poor generalisation.
- **Underfitting**: Model too simple → high bias, poor fit to training & test data.
- **Bias–Variance trade-off**:  
  - High bias → underfit.  
  - High variance → overfit.

### Methods to Improve Generalisation
- Cross-validation
- Regularisation (e.g., weight decay, priors in Bayesian models)
- Early stopping
- Bayesian model averaging

---

## 11.2 Calibration

**Goal:** Predicted probabilities should reflect actual frequencies.

Example: Of all predictions with $p=0.7$, about 70% should be correct.

### Calibration Metrics
- **Brier Score**:
$$
\mathrm{Brier} = \frac{1}{N} \sum_{i=1}^N (\hat{p}_i - y_i)^2
$$
Lower is better.

- **Expected Calibration Error (ECE)**:
  1. Bin predictions into $M$ bins.
  2. For each bin $B_m$, compute accuracy and average confidence.
  3. ECE = weighted average of absolute difference between accuracy and confidence:
$$
\mathrm{ECE} = \sum_{m=1}^M \frac{|B_m|}{N} \left| \mathrm{acc}(B_m) - \mathrm{conf}(B_m) \right|
$$

---

### Calibration Methods
- **Platt scaling**: Fit logistic regression to logits.
- **Isotonic regression**: Non-parametric, fits a monotonic mapping.
- **Temperature scaling**: Scales logits by a temperature $T$:
$$
\hat{p}_i = \mathrm{softmax}\left( \frac{z_i}{T} \right)
$$
$T>1$ smooths predictions.

## 11.3 Posterior Predictive Checks (PPC)

**Goal:** Evaluate model fit by comparing observed data to data simulated from the posterior predictive distribution:
$$
p(y^{\mathrm{rep}} \mid y) = \int p(y^{\mathrm{rep}} \mid \theta) \, p(\theta \mid y) \, d\theta
$$

**Procedure:**
1. Draw $\theta^{(s)} \sim p(\theta \mid y)$.
2. Simulate $y^{\mathrm{rep},(s)} \sim p(y \mid \theta^{(s)})$.
3. Compare $y^{\mathrm{rep}}$ to observed $y$ using test statistics or plots.

**Signs of misfit:**
- Systematic differences between replicated and observed data.
- Poor coverage of credible intervals.
