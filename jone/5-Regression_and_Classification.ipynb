# Section 5 — Regression & Classification

This section covers **Bayesian approaches** to regression and classification, focusing on common models like logistic regression.

---

## 5.1 Bayesian Regression

We model the relationship between inputs $\mathbf{x}$ and outputs $y$ by placing priors over parameters (or functions).

**Linear model:**
$$
y = \mathbf{x}^\top \mathbf{w} + \varepsilon,\quad \varepsilon \sim \mathcal{N}(0, \sigma^2)
$$

**Prior:**
$$
\mathbf{w} \sim \mathcal{N}(\mathbf{m}_0, S_0)
$$

**Posterior (conjugate):**
$$
S^{-1} = S_0^{-1} + \frac{1}{\sigma^2} X^\top X
$$
$$
\mathbf{m} = S \left(S_0^{-1} \mathbf{m}_0 + \frac{1}{\sigma^2} X^\top \mathbf{y} \right)
$$

**Predictive distribution** for new $\mathbf{x}^\star$:
$$
p(y^\star \mid \mathbf{x}^\star, \mathbf{X}, \mathbf{y}) =
\mathcal{N}\!\left(\mathbf{m}^\top \mathbf{x}^\star,\, \mathbf{x}^{\star\top} S \mathbf{x}^\star + \sigma^2\right)
$$

---

## 5.2 Bayesian Classification

Instead of predicting $y \in \mathbb{R}$, we predict **class probabilities**.

**Latent function view:**
- Introduce latent $f(\mathbf{x})$ with GP or linear prior.
- Apply a link function (sigmoid/softmax) to get class probabilities.
- Likelihood is **non-Gaussian** $\Rightarrow$ posterior intractable, need approximation (Laplace, VI, EP, MCMC).

---

## 5.3 Logistic Regression

**Binary case:**
$$
p(y=1 \mid \mathbf{x}, \mathbf{w}) = \sigma(\mathbf{x}^\top \mathbf{w}),
\quad \sigma(z) = \frac{1}{1 + e^{-z}}
$$
$$
p(y \mid \mathbf{x}, \mathbf{w}) = \sigma(\mathbf{x}^\top \mathbf{w})^y \,[1 - \sigma(\mathbf{x}^\top \mathbf{w})]^{1-y}
$$

**Prior:**
$$
\mathbf{w} \sim \mathcal{N}(\mathbf{0}, \alpha^{-1} I)
$$

**Posterior:**
$$
p(\mathbf{w} \mid \mathcal{D}) \propto p(\mathbf{y} \mid X, \mathbf{w})\, p(\mathbf{w})
$$
- No closed form; common approximations: Laplace around MAP, variational inference, MCMC.

**Predictive probability:**
$$
p(y^\star=1 \mid \mathbf{x}^\star, \mathcal{D}) = \int \sigma(\mathbf{x}^{\star\top} \mathbf{w})\, p(\mathbf{w} \mid \mathcal{D})\, d\mathbf{w}
$$

---

## 5.4 Logistic Sigmoid & Softmax

**Sigmoid function (binary classification):**
$$
\sigma(z) = \frac{1}{1 + e^{-z}}
$$
- Maps $\mathbb{R} \to (0,1)$.
- Symmetry: $\sigma(-z) = 1 - \sigma(z)$.
- Derivative: $\sigma'(z) = \sigma(z)\,[1 - \sigma(z)]$.

**Softmax function (multiclass classification):**
$$
\mathrm{softmax}(\mathbf{z})_k = \frac{\exp(z_k)}{\sum_{j=1}^K \exp(z_j)}
$$
- Maps $\mathbb{R}^K \to$ probability simplex.
- Invariant to constant shift: $\mathrm{softmax}(\mathbf{z} + c\mathbf{1}) = \mathrm{softmax}(\mathbf{z})$.

---

## 5.5 Exam Tips

- For Bayesian regression: Always write prior, likelihood, and posterior mean/cov.
- For classification: Emphasize **non-Gaussian likelihood** and approximation method.
- Logistic regression: state the link function, likelihood, and prior.
- Softmax: note invariance and use **log-sum-exp trick** for stability.
