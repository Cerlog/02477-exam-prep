# Section 12 — Hierarchical Bayesian Models

Hierarchical Bayesian models introduce **multiple levels of parameters**, allowing sharing of statistical strength across related groups.

---

## 12.1 Hierarchical Priors

In a standard Bayesian model:
$$
y_i \sim p(y_i \mid \theta), \quad \theta \sim p(\theta)
$$

In a **hierarchical model**:
$$
y_{ij} \sim p(y_{ij} \mid \theta_j), \quad \theta_j \sim p(\theta_j \mid \phi), \quad \phi \sim p(\phi)
$$
where:
- $\theta_j$ are **group-level parameters** for group $j$.
- $\phi$ are **hyperparameters** shared across groups.

**Benefits:**
- Borrow strength between groups.
- Avoid overfitting in small groups.
- Naturally represent multi-level structures.

---

## 12.2 Partial Pooling

**Pooling** refers to how information is shared across groups.

- **No pooling**: Fit each group separately (independent $\theta_j$).
- **Complete pooling**: Force all groups to share the same parameter $\theta$.
- **Partial pooling**: $\theta_j$ are similar but not identical, modeled via a common prior.

Example:  
$$
y_{ij} \sim \mathcal{N}(\theta_j, \sigma^2), \quad \theta_j \sim \mathcal{N}(\mu, \tau^2)
$$
where:
- $\mu$ is the population mean (estimated from all groups).
- $\tau^2$ controls the amount of pooling:
  - $\tau^2 \to 0$: complete pooling.
  - $\tau^2 \to \infty$: no pooling.

---

## 12.3 Inference in Hierarchical Models

- Can be done with:
  - Gibbs sampling (when conditionals are tractable).
  - HMC / NUTS.
  - Variational inference.

- Often requires sampling both group-level parameters $\theta_j$ and hyperparameters $\phi$.

---

## 12.4 Exam Tips

- Always draw the **plate diagram** to show hierarchy.
- Write down prior for $\theta_j$ and hyperprior for $\phi$.
- Emphasize the **benefits of partial pooling**.
- Remember: hierarchical priors shrink group estimates toward the population mean.
