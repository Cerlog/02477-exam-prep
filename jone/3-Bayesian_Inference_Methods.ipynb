# Section 3 — Bayesian Inference Methods

Bayesian inference often involves computing **posterior distributions** and **predictive distributions** that are analytically intractable.  
Here we cover **deterministic** and **stochastic** approximation methods, and diagnostics for sampling-based approaches.

---

## 3.1 Laplace Approximation

**Goal:** Approximate a complicated posterior $p(\theta\mid y)$ with a Gaussian centered at its mode (MAP).

### Procedure
1. Find $\hat{\theta}_{\mathrm{MAP}} = \arg\max_\theta \log p(\theta\mid y)$.
2. Compute the Hessian $H = \nabla^2_{\theta} \log p(\theta\mid y)$ at $\hat{\theta}_{\mathrm{MAP}}$.
3. Approximate:
$$
p(\theta\mid y) \approx \mathcal{N}\left(\hat{\theta}_{\mathrm{MAP}},\; -H^{-1}\right).
$$

**Notes:**
- Works best when posterior is **unimodal** and **peaked**.
- Useful for **marginal likelihood approximation**:
$$
p(y) \approx p(y\mid\hat\theta)p(\hat\theta)\,(2\pi)^{d/2}\,|{-H}^{-1}|^{1/2}.
$$

---

## 3.2 Grid Approximation

**Goal:** Approximate posterior by evaluating it over a discrete grid.

**Procedure:**
1. Choose parameter range(s) and create a grid.
2. Evaluate unnormalized posterior $p(y\mid\theta)p(\theta)$ at each grid point.
3. Normalize to sum/integrate to 1.
4. Approximate expectations by weighted sums.

**Pros:** Exact (within grid resolution).  
**Cons:** Curse of dimensionality — only feasible for low dimension ($\lesssim 2$–3).

---

## 3.3 Monte Carlo Sampling

### Standard Monte Carlo
Draw $\theta^{(s)} \overset{iid}{\sim} q(\theta)$, estimate expectations:
$$
\mathbb{E}_p[f(\theta)] \approx \frac{1}{S} \sum_{s=1}^S f(\theta^{(s)}).
$$

- If $q(\theta) = p(\theta\mid y)$, this is **direct sampling**.
- If $q(\theta) \neq p(\theta\mid y)$, use **importance sampling**:
$$
\mathbb{E}_p[f(\theta)] \approx \frac{\sum_{s=1}^S f(\theta^{(s)})\,w_s}{\sum_{s=1}^S w_s}, \quad w_s = \frac{p(\theta^{(s)})}{q(\theta^{(s)})}.
$$

---

### Gaussian–Hermite Quadrature
Used to approximate integrals of the form:
$$
\int_{-\infty}^\infty e^{-x^2} f(x)\,dx \;\approx\; \sum_{i=1}^n w_i\,f(x_i),
$$
where $(x_i, w_i)$ are **Hermite nodes and weights**.

- Common in probit models and latent Gaussian models.
- High accuracy for smooth $f(x)$.

---

### Probit Approximation
In Bayesian classification, the probit link $\Phi(\cdot)$ can be approximated for tractability:
$$
\Phi(x) \approx \sigma(\kappa x), \quad \kappa \approx \sqrt{\pi/8}
$$
where $\sigma$ is the logistic sigmoid.

---

## 3.4 Markov Chain Monte Carlo (MCMC)

**Goal:** Generate samples approximately from $p(\theta\mid y)$ when direct sampling is impossible.

### Metropolis–Hastings (MH)
1. Initialize $\theta^{(0)}$.
2. Propose $\theta' \sim q(\theta' \mid \theta^{(t)})$.
3. Accept with probability:
$$
\alpha = \min\left(1, \frac{p(y\mid\theta')p(\theta')\,q(\theta^{(t)}\mid\theta')}{p(y\mid\theta^{(t)})p(\theta^{(t)})\,q(\theta'\mid\theta^{(t)})} \right).
$$
4. If accepted, set $\theta^{(t+1)}=\theta'$, else $\theta^{(t+1)}=\theta^{(t)}$.
5. Repeat.

---

### Gibbs Sampling
Special case of MH where proposals are **full conditionals**:
$$
\theta_j^{(t+1)} \sim p(\theta_j \mid \theta_{-j}^{(t)}, y).
$$
- Always accepted.
- Requires that all full conditionals are easy to sample from.

---

### Ancestral Sampling
For directed graphical models:  
Sample variables in **topological order** from their conditional distributions.

Example: If $p(a,b,c) = p(a)\,p(b\mid a)\,p(c\mid a,b)$:
1. Sample $a \sim p(a)$.
2. Sample $b \sim p(b\mid a)$.
3. Sample $c \sim p(c\mid a,b)$.

---

### Rejection Sampling
1. Choose proposal $q(\theta)$ s.t. $p(\theta) \le M q(\theta)$ for some $M$.
2. Sample $\theta' \sim q(\theta)$ and $u \sim \mathrm{Uniform}(0,1)$.
3. Accept if $u \le \frac{p(\theta')}{M q(\theta')}$.

**Pros:** Simple, no correlations between samples.  
**Cons:** Inefficient if $M$ large.

## 3.5 MCMC Diagnostics

### R-hat (Gelman–Rubin Statistic)
- Compare **within-chain variance** to **between-chain variance**.
- $ \hat{R} \approx 1 $ means chains have mixed well.

### MCSE (Monte Carlo Standard Error)
- Estimate of uncertainty in the Monte Carlo estimate due to finite samples.
- For $\bar{f}$:
$$
\mathrm{MCSE}(\bar{f}) = \sqrt{\frac{\hat{\sigma}^2_{\mathrm{chain}}}{n_{\mathrm{eff}}}}.
$$

### Autocorrelation
- MCMC samples are correlated; autocorrelation measures persistence.
- High autocorrelation $\Rightarrow$ fewer **effective** samples.

### ESS (Effective Sample Size)
- Adjusted number of independent samples:
$$
\mathrm{ESS} \approx \frac{S}{1+2\sum_{k=1}^\infty \rho_k}
$$
where $\rho_k$ is lag-$k$ autocorrelation.

**Rule:** Higher ESS $\Rightarrow$ more reliable estimates.

---

## 3.6 Exam Tips

- For **Laplace**: write Taylor expansion of $\log p(\theta\mid y)$ and identify Hessian.  
- For **Grid Approximation**: mention normalization and limits in high-$d$.  
- For **MC**: distinguish between direct, importance sampling, quadrature.  
- For **MCMC**: always state acceptance rule and stationary target.  
- For **Diagnostics**: interpret R-hat $\approx 1$, high ESS, low autocorrelation as “good mixing”.
