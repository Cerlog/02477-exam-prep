# Section 1 — Bayesian Fundamentals

This section sets up the **language and core mechanics** of Bayesian inference you’ll use everywhere else.

---

## 1.1 Prior, Likelihood, Posterior, Posterior Predictive

- **Parameter**: unknown quantity $\theta$ (scalar/vector/matrix; sometimes a function).
- **Data**: observed sample $y$ (or $\mathbf{y}=(y_1,\dots,y_N)$).

### Prior $p(\theta)$
Your belief about $\theta$ **before** seeing the data.

> Examples:  
> $\theta\in[0,1]$: $\;\theta \sim \mathrm{Beta}(\alpha,\beta)$.  
> $\theta \in \mathbb{R}^d$: $\;\theta \sim \mathcal{N}(m_0,S_0)$.

### Likelihood $p(y\mid \theta)$
The data model: how likely is the observed $y$ given $\theta$.

> Examples:  
> Bernoulli/Binomial: $y \sim \mathrm{Bin}(N,\theta)$.  
> Gaussian noise: $y \sim \mathcal{N}(f(x;\theta), \sigma^2)$.

### Posterior $p(\theta\mid y)$
Updated belief after seeing data (via Bayes’ rule). Encodes **uncertainty** about $\theta$.
$$
p(\theta\mid y)\;=\;\frac{p(y\mid \theta)\,p(\theta)}{p(y)} \;\propto\; p(y\mid \theta)\,p(\theta).
$$

### Posterior Predictive $p(y^\star\mid y)$
Predictive distribution for **future** observation $y^\star$ that integrates over uncertainty in $\theta$:
$$
p(y^\star\mid y) \;=\; \int p(y^\star\mid \theta)\,p(\theta\mid y)\,d\theta.
$$

> **Intuition**: Don’t plug in a single $\theta$; **average** predictions across plausible $\theta$, weighted by the posterior.

---

## 1.2 Bayes’ Rule & Other Probability Rules

### Product/Chain Rule
$$
p(x,y)=p(x\mid y)\,p(y)=p(y\mid x)\,p(x).
$$

### Sum Rule (Marginalization)
- Continuous: $\;p(x)=\int p(x,z)\,dz=\int p(x\mid z)\,p(z)\,dz.$  
- Discrete: $\;p(x)=\sum_z p(x,z).$

### Bayes’ Rule
$$
p(\theta\mid y)=\frac{p(y\mid\theta)\,p(\theta)}{p(y)},
\qquad
p(y)=\int p(y\mid\theta)\,p(\theta)\,d\theta.
$$

### Law of Total Probability / Expectation
$$
p(y)=\int p(y\mid \theta)\,p(\theta)\,d\theta,
\qquad
\mathbb{E}[Y]=\mathbb{E}_\Theta\!\big[\mathbb{E}[Y\mid \Theta]\big].
$$

### Independence
$$
X\perp Y \iff p(x,y)=p(x)\,p(y) \iff p(x\mid y)=p(x).
$$

## 1.3 Plug-in Approximation

**Idea**: Replace the full posterior $p(\theta\mid y)$ by a **point estimate** $\hat\theta$ (e.g., MLE or MAP), then
$$
p(y^\star\mid y)\;\approx\;p(y^\star\mid \hat\theta)\quad\text{(no integration)}.
$$

- **Pros**: Simple, fast, often OK with large $N$ (posterior concentrates).  
- **Cons**: Discards parameter uncertainty $\Rightarrow$ too-narrow predictive intervals; can be **overconfident** with small/medium $N$.

> **Connection to Dirac delta** (below): Plug-in is equivalent to approximating the posterior by
> $p(\theta\mid y)\approx \delta(\theta-\hat\theta)$.

---

## 1.4 Dirac Delta $\delta(\cdot)$

A generalized function with the **sifting property**:
$$
\int f(\theta)\,\delta(\theta-a)\,d\theta \;=\; f(a).
$$

Key uses in Bayes:
- **Point mass** at $a$: $p(\theta)=\delta(\theta-a)$.  
- **Plug-in**: $p(\theta\mid y)\approx \delta(\theta-\hat\theta)$, hence
  $$
  p(y^\star\mid y)=\int p(y^\star\mid\theta)\,p(\theta\mid y)\,d\theta
  \approx \int p(y^\star\mid\theta)\,\delta(\theta-\hat\theta)\,d\theta
  = p(y^\star\mid \hat\theta).
  $$
- **Limit of Gaussians**: $\delta(\theta-a)=\lim_{\sigma^2\to 0}\,\mathcal{N}(\theta\mid a,\sigma^2)$.

## 1.5 Worked Examples

### (A) Beta–Binomial (Conjugate) — proportions
**Model**: $y\mid \theta \sim \mathrm{Bin}(N,\theta)$, prior $\theta\sim \mathrm{Beta}(\alpha,\beta)$.

- **Posterior**:
  $$
  \theta\mid y \;\sim\; \mathrm{Beta}(\alpha+y,\; \beta+N-y).
  $$
- **Posterior mean**: $\;\mathbb{E}[\theta\mid y]=\dfrac{\alpha+y}{\alpha+\beta+N}.$
- **Posterior predictive** for $y^\star$ successes in $N^\star$ trials:
  $$
  p(y^\star\mid y)=\binom{N^\star}{y^\star}\;
  \frac{B(\alpha+y+y^\star,\;\beta+N-y+N^\star-y^\star)}{B(\alpha+y,\;\beta+N-y)}.
  $$
- **Plug-in** with $\hat\theta$:
  $$
  p(y^\star\mid y)\approx \binom{N^\star}{y^\star}\,\hat\theta^{\,y^\star}(1-\hat\theta)^{N^\star-y^\star},
  \qquad
  \hat\theta_{\mathrm{MLE}}=\frac{y}{N},
  \quad
  \hat\theta_{\mathrm{MAP}}=\frac{\alpha+y-1}{\alpha+\beta+N-2}\;(\alpha,\beta>1).
  $$

> **Why Bayes helps**: with small $N$, MLE can be extreme (e.g., $0$ or $1$). The Beta posterior smooths estimates and yields calibrated predictive uncertainty.

---

### (B) Gaussian Mean (known variance) — simple analytic Bayes
**Model**: $y_1,\dots,y_N \overset{iid}{\sim}\mathcal{N}(\mu,\sigma^2)$ with known $\sigma^2$. Prior $\mu\sim \mathcal{N}(\mu_0,\tau_0^2)$.

- **Posterior**:
  $$
  \mu\mid \mathbf{y} \sim \mathcal{N}\!\big(\mu_N,\;\tau_N^2\big),
  \quad
  \tau_N^{-2}=\tau_0^{-2}+N\sigma^{-2},
  \quad
  \mu_N=\tau_N^2\!\left(\mu_0\tau_0^{-2} + \sigma^{-2}\sum_{n=1}^N y_n\right).
  $$
- **Posterior predictive** for a new $y^\star$:
  $$
  y^\star\mid \mathbf{y}\;\sim\;\mathcal{N}\big(\mu_N,\;\sigma^2+\tau_N^2\big)
  \quad\text{(data noise $+$ parameter uncertainty)}.
  $$
- **Plug-in** with $\hat\mu=\bar y$ (MLE) or $\hat\mu=\mu_N$ (MAP for Gaussian prior):
  $$
  y^\star\mid \mathbf{y}\;\approx\;\mathcal{N}\big(\hat\mu,\;\sigma^2\big)
  \quad\text{(misses the $+\tau_N^2$ term)}.
  $$

---

### (C) Bayesian Linear Regression (conjugate summary)
**Model**: $\mathbf{y}=\Phi \mathbf{w}+\boldsymbol\varepsilon,\ \boldsymbol\varepsilon\sim\mathcal{N}(0,\sigma^2 I)$; prior $\mathbf{w}\sim\mathcal{N}(\mathbf{m}_0,S_0)$.

- **Posterior**:
  $$
  S^{-1} \;=\; S_0^{-1} + \frac{1}{\sigma^2}\,\Phi^\top \Phi,
  \qquad
  \mathbf{m} \;=\; S\!\left(S_0^{-1}\mathbf{m}_0 + \frac{1}{\sigma^2}\,\Phi^\top \mathbf{y}\right).
  $$
- **Predictive** at $\phi^\star$:
  $$
  y^\star\mid \mathbf{y}\;\sim\;\mathcal{N}\!\Big(\mathbf{m}^\top \phi^\star,\;\;\phi^{\star\top}S\phi^\star + \sigma^2\Big).
  $$
- **Plug-in** with $\hat{\mathbf{w}}$:
  $$
  y^\star\mid \mathbf{y}\;\approx\; \mathcal{N}\big(\hat{\mathbf{w}}^\top \phi^\star,\,\sigma^2\big).
  $$

## 1.6 Practical Checklist (what to write fast in the exam)

1. **State the model** (prior + likelihood) clearly.  
2. **Write Bayes’ rule**: $p(\theta\mid y)\propto p(y\mid\theta)\,p(\theta)$.  
3. If **conjugate**, give the **posterior family & parameters**.  
4. For predictions, write the **posterior predictive** and then:
   - Conjugate case: give **closed form** (e.g., Beta–Binomial, Gaussian predictive).  
   - Otherwise: state **approximation** (plug-in, Laplace, grid, Monte Carlo).  
5. If using **plug-in**, explicitly note it corresponds to $p(\theta\mid y)\approx \delta(\theta-\hat\theta)$.  
6. Mention **uncertainty decomposition** in predictive: *data noise $+$ parameter uncertainty* (when applicable).
