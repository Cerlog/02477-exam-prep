# Section 9 — Bayesian Neural Networks (BNNs)

Bayesian Neural Networks place **probability distributions** over the weights of a neural network, rather than using fixed point estimates.

---

## 9.1 BNN Basics

A standard neural network predicts:
$$
\hat{y} = f_{\mathbf{w}}(\mathbf{x})
$$
with **deterministic** weights $\mathbf{w}$.

A **Bayesian Neural Network** instead models:
$$
p(\mathbf{w}) \quad \text{(prior over weights)}
$$
$$
p(\mathbf{y} \mid \mathbf{x}, \mathbf{w}) \quad \text{(likelihood)}
$$
and infers the posterior:
$$
p(\mathbf{w} \mid \mathcal{D}) \propto p(\mathbf{y} \mid \mathbf{X}, \mathbf{w}) \, p(\mathbf{w})
$$

Predictions marginalize over the weight posterior:
$$
p(y^\star \mid x^\star, \mathcal{D}) = \int p(y^\star \mid x^\star, \mathbf{w}) \, p(\mathbf{w} \mid \mathcal{D}) \, d\mathbf{w}
$$

---

## 9.2 Priors for BNN Weights

- **Gaussian prior**: $\mathcal{N}(0, \sigma_w^2 I)$ — encourages small weights.
- **Laplace prior**: promotes sparsity.
- Priors can encode **smoothness**, **scale**, or other inductive biases.

---

## 9.3 Posterior Inference in BNNs

Exact inference is intractable because:
- High-dimensional parameter space.
- Nonlinear likelihood.

**Approximation methods:**
- **MCMC**: e.g., Hamiltonian Monte Carlo (HMC), Stochastic Gradient Langevin Dynamics (SGLD)
- **Variational Inference**: e.g., mean-field Gaussian $q(\mathbf{w})$
- **Laplace approximation** around MAP weights.

---

## 9.4 Variational Inference for BNNs

Assume:
$$
q_\lambda(\mathbf{w}) = \prod_j \mathcal{N}(w_j \mid \mu_j, \sigma_j^2)
$$
Maximize ELBO:
$$
\mathcal{L}(\lambda) = \mathbb{E}_{q_\lambda(\mathbf{w})}[\log p(\mathbf{y} \mid \mathbf{X}, \mathbf{w})] - \mathrm{KL}(q_\lambda(\mathbf{w}) \,\|\, p(\mathbf{w}))
$$

**Reparameterisation trick**:
$$
w_j = \mu_j + \sigma_j \epsilon_j, \quad \epsilon_j \sim \mathcal{N}(0,1)
$$

---

## 9.5 Prediction with BNNs

Monte Carlo estimation:
1. Sample $\mathbf{w}^{(s)} \sim p(\mathbf{w} \mid \mathcal{D})$ (or $q_\lambda$ in VI).
2. Compute $p(y^\star \mid x^\star, \mathbf{w}^{(s)})$.
3. Average over samples.

This yields:
- **Predictive mean**: expected output.
- **Predictive uncertainty**: variance across weight samples.

---

## 9.6 Benefits & Challenges

**Benefits:**
- Uncertainty quantification in predictions.
- Regularization through priors.
- Principled handling of small data.

**Challenges:**
- Computationally expensive.
- Complex posterior distributions.

---

## 9.7 Exam Tips

- Always start by writing **Bayes’ rule over weights**.
- For VI: mention mean-field assumption and reparameterisation trick.
- For MCMC: mention gradient-based samplers for high-dimensional $\mathbf{w}$.
- Highlight difference from **dropout as approximate Bayesian inference** (Gal & Ghahramani, 2016).
