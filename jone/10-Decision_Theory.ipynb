# Section 10 — Decision Theory

Decision theory formalizes how to make **optimal decisions** under uncertainty, using probability models and a utility (or loss) function.

---

## 10.1 Decision Theory Basics

We want to choose an **action** $a \in \mathcal{A}$ based on observed data $y$ and an uncertain state $\theta$.

- **Loss function**: $L(a, \theta)$ measures the cost of taking action $a$ when the true state is $\theta$.
- **Utility function**: $U(a, \theta) = -L(a, \theta)$ is the reward.

Our **goal**: choose $a$ to minimize expected loss (or maximize expected utility).

---

### 10.1.1 Frequentist Risk

Under frequentist decision theory, the **risk** is:
$$
R(\theta, a) = \mathbb{E}_{y \mid \theta} \big[ L(a(y), \theta) \big]
$$
We might minimize **maximum risk** (minimax rule).

---

### 10.1.2 Bayesian Decision Rule

In Bayesian decision theory, we treat $\theta$ as random and minimize the **posterior expected loss**:
$$
a^\star(y) = \arg\min_a \mathbb{E}_{\theta \mid y} \big[ L(a, \theta) \big]
$$

---

## 10.2 Common Loss Functions & Optimal Decisions

- **0–1 Loss** (classification):
$$
L(a, \theta) =
\begin{cases}
0, & a = \theta \\
1, & a \neq \theta
\end{cases}
$$
Bayes rule: choose **MAP estimate**:
$$
a^\star(y) = \arg\max_\theta p(\theta \mid y)
$$

- **Squared Error Loss** (regression):
$$
L(a, \theta) = (a - \theta)^2
$$
Bayes rule: choose **posterior mean**:
$$
a^\star(y) = \mathbb{E}[\theta \mid y]
$$

- **Absolute Error Loss**:
$$
L(a, \theta) = |a - \theta|
$$
Bayes rule: choose **posterior median**.

---

## 10.3 Applications

- **Classification**: choose class with highest posterior probability (MAP).
- **Regression**: choose posterior mean or median, depending on loss.
- **Medical decision-making**: weigh benefits vs risks via utilities.
- **Economics**: optimize expected profit or minimize expected cost.

---

## 10.4 Exam Tips

- Always start by **defining loss function** and expected loss.
- For Bayesian decision rules:  
  1. Write posterior $p(\theta \mid y)$.  
  2. Write expected loss over posterior.  
  3. Find $a^\star(y)$ that minimizes it.
- Know **which posterior summary** corresponds to each loss function (MAP, mean, median).
