# Section 6 — Estimation & Optimization

This section covers **point estimation** methods and common optimization tools used in Bayesian and frequentist contexts.

---

## 6.1 Maximum Likelihood Estimator (MLE)

**Goal:** Find the parameter $\hat{\theta}_{\mathrm{MLE}}$ that maximizes the likelihood:
$$
\hat{\theta}_{\mathrm{MLE}} = \arg\max_{\theta} p(\mathbf{y} \mid \theta)
$$

Equivalent to maximizing the log-likelihood:
$$
\hat{\theta}_{\mathrm{MLE}} = \arg\max_{\theta} \; \ell(\theta) = \log p(\mathbf{y} \mid \theta)
$$

**Properties:**
- Asymptotically unbiased and efficient.
- Ignores prior information.
- Can be **overconfident** with small datasets.

---

## 6.2 Maximum A Posteriori (MAP) Estimation

**Goal:** Find the mode of the posterior distribution:
$$
\hat{\theta}_{\mathrm{MAP}} = \arg\max_{\theta} p(\theta \mid \mathbf{y})
$$

Using Bayes’ rule:
$$
\hat{\theta}_{\mathrm{MAP}} = \arg\max_{\theta} \big[ \log p(\mathbf{y} \mid \theta) + \log p(\theta) \big]
$$

**Special cases:**
- If prior is uniform, MAP = MLE.
- For Gaussian prior on weights, MAP = ridge regression.

---

## 6.3 Score Function

The **score function** is the gradient of the log-likelihood:
$$
U(\theta) = \nabla_\theta \log p(\mathbf{y} \mid \theta)
$$

**Properties:**
- $\mathbb{E}[U(\theta)] = 0$ under the model.
- Used in:
  - Newton–Raphson for MLE
  - Fisher scoring
  - Variance estimation (via Fisher information):
    $$
    I(\theta) = -\mathbb{E}\left[ \nabla^2_\theta \log p(\mathbf{y} \mid \theta) \right] = \mathrm{Var}[U(\theta)]
    $$

---

## 6.4 Stochastic Gradient Descent (SGD)

**Goal:** Minimize a loss function $\mathcal{L}(\theta)$ using noisy gradient estimates.

**Update rule:**
$$
\theta^{(t+1)} = \theta^{(t)} - \eta_t \,\widehat{\nabla_\theta \mathcal{L}(\theta^{(t)})}
$$
where:
- $\eta_t$ = learning rate at iteration $t$
- Gradient estimated from a **mini-batch** of data.

**Properties:**
- Efficient for large datasets.
- Can be combined with momentum, Adam, RMSprop, etc.
- In Bayesian VI: SGD is used to optimize ELBO.

---

## 6.5 Exam Tips

- MLE: Write $\hat{\theta} = \arg\max_\theta \log p(y\mid\theta)$ and show derivative = 0.
- MAP: Add $\log p(\theta)$ term to MLE derivation.
- Score function: Remember its **expectation is zero**.
- SGD: Mention stochastic estimate of gradient and learning rate schedule.
