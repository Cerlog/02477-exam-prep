```python
import jax
import jax.numpy as jnp
from jax import random, jit
from jax.scipy.special import ndtr

mu = jnp.array([0.0, 0.6])
Sigma = jnp.array([[1.0, 0.235],
                   [0.235, 1.0]])
```

```python
def analytic_prob_w1_gt_w2(mu, Sigma):
    mean = mu[0] - mu[1]
    var  = Sigma[0,0] + Sigma[1,1] - 2.0*Sigma[0,1]
    return ndtr(mean / jnp.sqrt(var))

def mc_prob_w1_gt_w2(key, mu, Sigma, S: int):
    samples = random.multivariate_normal(key, mean=mu, cov=Sigma, shape=(S,))
    ind = (samples[:, 0] > samples[:, 1]).astype(jnp.float32)
    p_hat = ind.mean()
    mcse  = jnp.sqrt(p_hat * (1.0 - p_hat) / S)
    return p_hat, mcse

key = random.PRNGKey(42)
S = 500
key, subkey = random.split(key)
p_hat, mcse = mc_prob_w1_gt_w2(subkey, mu, Sigma, S)
p_exact = analytic_prob_w1_gt_w2(mu, Sigma)

print(f"Analytic:    {p_exact:.3f}")
print(f"MC estimate: {p_hat:.3f}  (MCSE ≈ {mcse:.3f})")
print(f"~95% MC interval: [{p_hat - 1.96*mcse:.3f}, {p_hat + 1.96*mcse:.3f}]")
```
