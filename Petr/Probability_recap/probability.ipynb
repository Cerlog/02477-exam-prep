```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
class FruitBoxExperiment:
    def __init__(self):
        # define the setup 
        self.red_box = {'apples': 2, 'oranges': 6}
        self.blue_box = {'apples':3, 'oranges': 1}
        self.box_probs = {'red': 0.4, 'blue': 0.6}
        

    def calculate_conditional_prob(self, fruit, box):
        """Calculate P(fruit | box)"""
        if box == 'red':
            total = self.red_box['apples'] + self.red_box['oranges']
            return self.red_box[fruit] / total if total > 0 else 0
        elif box == 'blue':
            total = self.blue_box['apples'] + self.blue_box['oranges']
            return self.blue_box[fruit] / total if total > 0 else 0
    
    def calculate_joint_prob(self, fruit, box):
        """Calculate P(fruit, box)"""
        return self.box_probs[box] * self.calculate_conditional_prob(fruit, box)
    
    def calculate_marginal_prob(self, fruit):
        """Calculate P(fruit)"""
        return (self.calculate_joint_prob(fruit, 'red') + 
                self.calculate_joint_prob(fruit, 'blue'))
        
exp = FruitBoxExperiment()
print(f"P(apple) = {exp.calculate_marginal_prob('apples'):.3f}")
print(f"P(orange) = {exp.calculate_marginal_prob('oranges'):.3f}")
print(f"P(apple | red box) = {exp.calculate_conditional_prob('apples', 'red'):.3f}")
print(f"P(orange | blue box) = {exp.calculate_conditional_prob('oranges', 'blue'):.3f}")
print(f"P(apple, red box) = {exp.calculate_joint_prob('apples', 'red'):.3f}")
print(f"P(orange, blue box) = {exp.calculate_joint_prob('oranges', 'blue'):.3f}")
print(f"P(apple, blue box) = {exp.calculate_joint_prob('apples', 'blue'):.3f}")
print(f"P(orange, red box) = {exp.calculate_joint_prob('oranges', 'red'):.3f}"  )


joint_probs = [
    exp.calculate_joint_prob('apples', 'red'),
    exp.calculate_joint_prob('oranges', 'red'),
    exp.calculate_joint_prob('apples', 'blue'),
    exp.calculate_joint_prob('oranges', 'blue')]

print("*" * 50)

print(f"Sum of the joint probaiblities is {np.sum(joint_probs)}")
```
