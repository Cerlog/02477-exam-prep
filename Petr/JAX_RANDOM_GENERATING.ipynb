# JAX RANDOM NUMBER GENERATION

Jax follows the functional programming paradigm, which means that it does not have a global state. This is important for reproducibility and parallelism. In Jax, random number generation is handled through a functional interface that requires you to pass around a random key.

# NUMPY RANDOM NUMBER GENERATION

Numpy has seeding where internally it uses a global state to generate random numbers. This means that if you set the seed, you will get the same random numbers every time you run the code. This is useful for reproducibility, but it can also lead to issues if you are not careful.



```python
import numpy as np 
np.random.seed(0)
```

```python
np.random.random()
```

# JAX RANDOM NUMBER GENERATION

Jax focuses on 3 main functions for random number generation:

- reproducibility 
- parallelism
- vectorization 


jax us to manage the random state explicitly. This means that you need to pass around a random key, which is a unique identifier for the random state. This key is used to generate random numbers and can be split to create new keys for parallel operations.

in jax we must manage nad pass the key when we generate random numbers 

if we use the same key, we get the same value (deterministic)

```python
from jax import random 

key = random.key(42)
print(key)
```

```python
random.normal(key)
```

```python
new_key, subkey = random.split(key) 
del key # the old key is consumed by split() - we must never use it again 

val = random.normal(subkey)

val
```
