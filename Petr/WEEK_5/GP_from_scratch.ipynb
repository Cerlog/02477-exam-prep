```python
sigma_2 = 1 
ell = 1 

def squared_exponential_kernel(x, ell=1, sigma2=1.0):
    """
    Compute the squared exponential (RBF) kernel matrix.
    
    This kernel computes:
        K_ij = sigma2 * exp( - (x_i - x_j)^2 / (2 * ell^2) )
        
    Args:
        x (jnp.ndarray): 1D array of input points, shape (n,)
        ell (float): Length-scale parameter. Controls smoothness. Default is 1.0.
        sigma2 (float): Output variance. Scales the kernel. Default is 1.0.
        
    Returns:
        jnp.ndarray: Squared exponential kernel matrix of shape (n, n)
        
    Example:
        >>> import jax.numpy as jnp
        >>> x = jnp.array([-1.0, 0.0, 1.0])
        >>> K = squared_exponential_kernel(x)
        >>> print(K)
        [[1.         0.60653067 0.13533528]
        [0.60653067 1.         0.60653067]
        [0.13533528 0.60653067 1.        ]]
    """
    
    # reshape x into column and row vectors 
    x_col = x.reshape(-1, 1)    # shape (n, 1)
    x_row = x.reshape(1, -1)    # shape (1, n)
    
    # compute squared pairwise distances 
    sq_dist = (x_col - x_row) ** 2  # shape (n, n)
    
    # compute kernel matrix 
    K = sigma2 * jnp.exp(-sq_dist / (2 * ell ** 2)) # shape (n, n)
    
    return K 
```
